```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Your tasks with urgency and importance scores (0-10 scale)
tasks = [
    {"name": "SQL-learning", "urgency": 8, "importance": 7},
    {"name": "LPIC-review", "urgency": 4, "importance": 6}
]

# Create figure
fig, ax = plt.subplots(figsize=(10, 8))

# Define quadrant colors
colors = {
    'Q1': '#ffcccc',  # Urgent & Important (red)
    'Q2': '#ccddff',  # Not Urgent but Important (blue)
    'Q3': '#fff4cc',  # Urgent but Not Important (yellow)
    'Q4': '#e6e6e6'   # Not Urgent & Not Important (gray)
}

# Draw quadrants
quadrants = [
    patches.Rectangle((5, 5), 5, 5, facecolor=colors['Q1'], alpha=0.3),  # Top-right
    patches.Rectangle((0, 5), 5, 5, facecolor=colors['Q2'], alpha=0.3),  # Top-left
    patches.Rectangle((5, 0), 5, 5, facecolor=colors['Q3'], alpha=0.3),  # Bottom-right
    patches.Rectangle((0, 0), 5, 5, facecolor=colors['Q4'], alpha=0.3),  # Bottom-left
]

for quadrant in quadrants:
    ax.add_patch(quadrant)

# Plot tasks
for task in tasks:
    ax.scatter(task['urgency'], task['importance'], s=200, alpha=0.6, edgecolors='black')
    ax.annotate(task['name'], (task['urgency'], task['importance']), 
                ha='center', va='center', fontsize=12, weight='bold')

# Add quadrant labels
ax.text(2.5, 7.5, 'SCHEDULE\n(Do Later)', ha='center', va='center', 
        fontsize=12, weight='bold', alpha=0.7)
ax.text(7.5, 7.5, 'DO FIRST\n(Do Now)', ha='center', va='center', 
        fontsize=12, weight='bold', alpha=0.7)
ax.text(7.5, 2.5, 'DELEGATE', ha='center', va='center', 
        fontsize=12, weight='bold', alpha=0.7)
ax.text(2.5, 2.5, 'ELIMINATE', ha='center', va='center', 
        fontsize=12, weight='bold', alpha=0.7)

# Add grid lines at middle
ax.axhline(y=5, color='black', linewidth=2)
ax.axvline(x=5, color='black', linewidth=2)

# Labels and formatting
ax.set_xlabel('Urgency →', fontsize=14, weight='bold')
ax.set_ylabel('Importance →', fontsize=14, weight='bold')
ax.set_title('Eisenhower Matrix - Task Prioritization', fontsize=16, weight='bold', pad=20)
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```
